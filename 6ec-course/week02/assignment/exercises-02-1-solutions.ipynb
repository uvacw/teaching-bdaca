### Task 2a — Complete a function

Replace `pass` with an `if/else` statement. Return `"long"` for 500 words or more, and `"short"` otherwise.

Run the three calls. Discuss why testing exactly 500 is useful.

```python
def article_length(word_count):
    if word_count >= 500:
        return "long"
    else:
        return "short"

print(article_length(450))
print(article_length(500))
print(article_length(900))
```

**Check:** `short`, `long`, `long`. Testing exactly 500 checks the boundary: `>=` includes 500 in the long category, whereas `>` would incorrectly classify it as short.

### Task 2b — Use a loop

Loop over `word_counts`. For each value, print the count and the result of `article_length()`.

**Hint:** start with `for count in word_counts:`.

```python
for count in word_counts:
    print(count, article_length(count))
```

### Task 5b — Select transport articles

Complete the loop: if `topic` is `"transport"`, add the `article_id` to `transport_ids`.

### Task 6a — Read a CSV

Run the cell below. `csv.DictReader(f)` reads each row as a dictionary; the loop stores the rows in a list.

```python
# PROVIDED — run as written.
csv_rows = []

with open(f"{folder}/article_summary.csv", mode="r", encoding="utf-8", newline="") as f:
    for row in csv.DictReader(f):
        csv_rows.append(row)

print(len(csv_rows))
print(csv_rows[0])
```

**Your code:** print the first row's `word_count` and its type. Convert it with `int()`, then print the converted value and its type.

```python
csv_count = csv_rows[0]["word_count"]
print(csv_count, type(csv_count))

numeric_count = int(csv_count)
print(numeric_count, type(numeric_count))
```
