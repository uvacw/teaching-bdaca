```python
## example creating a dict with information on source / data / title:

from glob import glob
import os

data = []

for fn in glob("articles/**/*", recursive=True):
    if os.path.isfile(fn):  # skip folders
        parts = fn.split(os.sep)  # os.sep handles "/" or "\" depending on OS
        
        date = parts[1] if len(parts) > 1 else None
        source = parts[3] if len(parts) > 3 else None

        title = os.path.basename(fn)
        with open(fn, mode='r', encoding='utf-8', errors='ignore') as fi:
            text = fi.read()

        # Store structured data
        data.append({
            "date": date,
            "source": source,
            "title": title,
            "text": text
        })

print(f"Loaded {len(data)} articles.")
```
